1. Text Prompt

A Text Prompt is a prompt that generates a single block of text (string) to send to the language model. It is used when the model only needs plain text as input and does not require conversation roles or chat history.

2. Message Prompt (Chat Prompt)

A Message Prompt is a prompt that generates a list of messages, where each message has a specific role (such as system, human, or AI). It is designed for chat-based language models to provide structured conversations and maintain context.

Types of Message Prompts
1. System Message Prompt

Defines the AI assistant's behavior, role, personality, or instructions before the conversation begins.

2. Human Message Prompt

Represents the user's input or question that is sent to the chat model.

3. AI Message Prompt

Represents a previous assistant response. It is commonly used to provide examples or continue an existing conversation.

4. Messages Placeholder

A placeholder that dynamically inserts one or more existing messages (such as conversation history) into the prompt, allowing the model to maintain context across interactions.

```python
import os
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
load_dotenv()

os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")

model=init_chat_model("groq:qwen/qwen3-32b")
```

Text Prompt

```python
model.invoke("What is Docker?")
```

```python
from langchain.messages import AIMessage,ToolMessage,SystemMessage,HumanMessage

message=[
    SystemMessage("You are a Senior Devops Engineer"),
    HumanMessage("What is the diff between docker and kubernetes?")
]

response = model.invoke(message)
response.content
```

```python
# Message Metadata
human_msg = HumanMessage(
    content = "HELLO",
    name = "jawad", #identify diff users
    id = "msg_123"  #to track the msg
)
res = model.invoke([human_msg])
res.content
```
