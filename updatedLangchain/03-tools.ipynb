Tool

Tool: A tool is an external function, API, software, or service that an AI agent uses to perform tasks it cannot accomplish using the language model alone.

Why Tools are Used

Purpose of Tools: Tools enable AI agents to interact with the real world, retrieve up-to-date information, execute code, access databases, and automate tasks.

Common AI Agent Tools
1. Web Search Tool

A tool that retrieves the latest information from the internet.

2. Calculator Tool

A tool that performs accurate mathematical calculations.

3. Python Tool

A tool that executes Python code for data analysis, automation, and computations.

4. File Tool

A tool that reads, writes, creates, and modifies files such as PDFs, Word documents, Excel sheets, and CSV files.

5. Database Tool

A tool that queries, retrieves, updates, or deletes information stored in a database.

6. API Tool

A tool that communicates with external applications or services through Application Programming Interfaces (APIs).

7. Browser Tool

A tool that visits websites, extracts information, and interacts with web pages.

8. Email Tool

A tool that sends, receives, and manages email messages.

9. Calendar Tool

A tool that creates, updates, and manages events and schedules.

10. Image Generation Tool

A tool that generates images from text descriptions or edits existing images.

11. Terminal/Shell Tool

A tool that executes operating system commands to automate system tasks.

12. Retrieval (RAG) Tool

A tool that searches documents or a knowledge base to retrieve relevant information before generating a response.

13. GitHub Tool

A tool that interacts with GitHub repositories to create commits, pull requests, issues, and manage code.

14. Slack Tool

A tool that sends, reads, and manages messages in Slack workspaces.

15. Translation Tool

A tool that translates text between different languages.

16. Weather Tool

A tool that retrieves real-time weather information for a specified location.

17. Maps/Location Tool

A tool that provides navigation, directions, nearby places, and geographic information.

Tool Calling

Tool Calling: Tool calling is the process in which an AI agent decides to invoke an external tool, passes the required inputs, receives the output, and uses the result to complete the user's request.

Function Calling

Function Calling: Function calling is a mechanism that allows a language model to invoke predefined functions with structured arguments to perform specific tasks.

```python
import os
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
load_dotenv()

os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")

model=init_chat_model("groq:qwen/qwen3-32b")
response=model.invoke("what is a dinasaur?")
# response.content
response
```

```python
from langchain.tools import tool

@tool
def getWeather(location:str)->str:
    """Get weather of the location"""
    print("The weather is hot ",location)

model_with_tools=model.bind_tools([getWeather])
```

```python
response = model_with_tools.invoke("Whts the weather in bostn ?")
print(response)
print(response.content)

for tool_call in response.tool_calls:
    print("Tool : ",tool_call['name'])
    print("Args : ",tool_call['args'])
```
