### Model Integeration with OpenAI,Google Gemini,Groq

```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["GOOGLE_API_KEY"]=os.getenv("GOOGLE_API_KEY")
os.environ["OPENAI_API_KEY"]=os.getenv("OPENAI_API_KEY")
os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")
```

Gemini Integeration

```python
from langchain.chat_models import init_chat_model

model=init_chat_model("google_genai:gemini-2.5-flash-lite")
model
```

```python
response=model.invoke("what is a dinasaur?")
response.content
# response
```

```python
import os
from langchain_google_genai import ChatGoogleGenerativeAI

model = ChatGoogleGenerativeAI(model="gemini-2.5-flash-lite")
response = model.invoke("hey! how are yiu?")
response.content
```

Groq Integeration

```python
from langchain.chat_models import init_chat_model

model=init_chat_model("groq:qwen/qwen3-32b")
response=model.invoke("what is a dinasaur?")
# response.content
response
```

Streaming

```python
for chunk in model.stream("why do parrots have colorful feathers"):
    print(chunk.text, end="|", flush=True)
```

```python
responses = model.batch([
    "what is computer language?",
    "what is python?",
    "what is high level computer language?"
])
for response in responses:
    print(response.content)
```

```python
responses = model.batch(
    ["what is computer language?",
    "what is python?",
    "what is high level computer language?"],
    config = {
        'max_concurrency' : 5,
        # if 10 questions are given at a time it will try 5 first and then other 5.
    }
)
for response in responses:
    print(response.content)
```
